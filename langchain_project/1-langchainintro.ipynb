Streaming and Batch

Streaming

Most model can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experiance, particularly for longer response. Celling stream() returns an itrator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time.

```python
model.invoke("Write me 500 words paragraph on AI")
```

```python
for chunk in model.stream("Write me 500 words paragraph on AI"):
    print(chunk.text, end="|", flush=True)
```

Batch

Batching a collection of independent requests to a model can significantly improve performance and reduce cost, as the processing can be done in paralled:

```python
responses=model.batch(
    [
        "Why do parrots have colorful feathers?",
        "How do airplanes fly?",
        "What is quantum computing?"
    ]
)
for response in responses:
    print(response)
```
